```python
from pathlib import Path
import subprocess

repo_root = Path.cwd().resolve()
python_executable = repo_root / ".venv" / "Scripts" / "python.exe"
entry_point = repo_root / "rhi_robotics" / "run_rhi_loop.py"

if not python_executable.is_file():
    raise FileNotFoundError(f"Virtual environment Python not found: {python_executable}")
if not entry_point.is_file():
    raise FileNotFoundError(f"RHI entry point not found: {entry_point}")

process = subprocess.Popen(
    [str(python_executable), "-m", "rhi_robotics.run_rhi_loop"],
    cwd=repo_root,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    encoding="utf-8",
    errors="replace",
    bufsize=1,
)

if process.stdout is None:
    raise RuntimeError("Could not capture RHI process output.")

for output_line in process.stdout:
    print(output_line, end="", flush=True)

return_code = process.wait()
if return_code != 0:
    raise RuntimeError(
        f"RHI process exited with status {return_code}. "
        "Review the error output above for the actual cause."
    )
```
